```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("retail_business_analysis.csv")

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nMissing Values:")
print(df.isnull().sum())

total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
total_cost = df["Cost"].sum()

print("\nTotal Sales:", round(total_sales, 2))
print("Total Cost:", round(total_cost, 2))
print("Total Profit:", round(total_profit, 2))
```

```python
# Category-wise Sales and Profit

category = df.groupby("Category")[["Sales", "Profit"]].sum().sort_values("Profit", ascending=False)

print(category.round(2))

category.plot(kind="bar", figsize=(8,5))
plt.title("Category-wise Sales and Profit")
plt.xlabel("Category")
plt.ylabel("Amount")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Region-wise Sales and Profit

region = df.groupby("Region")[["Sales", "Profit"]].sum().sort_values("Profit", ascending=False)

print(region.round(2))

region.plot(kind="bar", figsize=(8,5))
plt.title("Region-wise Sales and Profit")
plt.xlabel("Region")
plt.ylabel("Amount")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Inventory Days vs Profit

correlation = df["Inventory_Days"].corr(df["Profit"])

print("Correlation between Inventory Days and Profit:", round(correlation, 3))

plt.figure(figsize=(8,5))
plt.scatter(df["Inventory_Days"], df["Profit"])
plt.title("Inventory Days vs Profit")
plt.xlabel("Inventory Days")
plt.ylabel("Profit")
plt.tight_layout()
plt.show()
```
